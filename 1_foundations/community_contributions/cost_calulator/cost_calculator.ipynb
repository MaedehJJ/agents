## Welcome to the Cost Calulator App,

Using this app, you can have an estimation of the time, money and people you need to get a project done.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
```

```python

openai_api_key = os.getenv('OPENAI_API_KEY')
google_api_key = os.getenv('GOOGLE_API_KEY')
groq_api_key = os.getenv('GROQ_API_KEY')

if openai_api_key and google_api_key and groq_api_key:
    print(f"All API Keys exist")
else:
    print("One or more API Keys are not set")
```

```python
# Getting the project description from the user

project_description = input("Enter project description: ").strip()
```

```python
request = """Given the project description, break the project into clear, actionable components as follows:
Phases: Identify high-level phases of the project (e.g., planning, design, development, testing, etc.).
Milestones: Break each phase down into specific milestones that represent significant progress points or deliverables.
Tasks: Break each milestone down further into tasks that are specific and actionable. Each task should be a step toward achieving the milestone.
Dependencies: Identify any tasks or milestones that are dependent on the completion of others.
Estimated Time: Provide an estimated time for completion for each task and milestone (e.g., hours, days, weeks).
Human Resource Requirements: Identify the types of human resources required for each task, including skill sets and number of people.
Tools and Technologies: Identify any tools, software, or technologies required for the tasks.
Resources and Materials: Identify any non-human resources (e.g., equipment, materials) needed for the tasks.
Once you've broken the project down, create a clear and structured prompt for Agent B that includes the following:
Complete List of Tasks: Include all tasks with their time estimates, required resources, and dependencies.
Human Resource Requirements: Specify the skill sets, number of people, and time each resource will need to dedicate to each task.
Tools/Technologies: Include any tools or technologies needed for specific tasks.
Milestone Overview: List key milestones with their associated tasks and time estimates.
In the prompt for Agent B, make sure to provide clear instructions to estimate the cost in terms of money, human resources, and time based on the provided task breakdown and resource requirements."""

request += f"User's project description: {project_description}"
messages = [{"role": "user", "content": request}]

openai = OpenAI()
response = openai.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages,
)
initial_prompt = response.choices[0].message.content
display(Markdown(initial_prompt))
```

```python
messages = [{"role": "user", "content": initial_prompt}]
final_cost = openai.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages,
)
final_cost = final_cost.choices[0].message.content
display(Markdown(final_cost))
```
